# Simulated Galaxies

We can learn a lot from observations/measurements of galaxies (as you saw in the last notebook). However, we also talked about how galaxies can be oriented in a certain direction to us, which limits our view of them. We said galaxies change over billions of years; even mergers are slow. And, we also said that some galaxies are too far away for us to see. These are all pretty big limitations!

Luckily for astronomers, we understand how they work pretty well (but not perfectly). We understand enough that we can tell a computer how galaxies work and then have the computer make, or simulate, a galaxy. The nice thing about having a computer simulate a galaxy is then we can watch the galaxy's entire life. We can also run experiments that we cannot do in real life. This allows us to watch mergers happen.

```python
#import our graphing package
# Import packages
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib
matplotlib.rc('font', **{'family': 'sans-serif', 'size': 16})
```

```python
#import numpy
import numpy as np

# Load the data
data1 = pd.read_csv('../Data/individual_analogs/453134_star_properties.csv')

# Calculate color (g - r)
color = data1['g band'] - data1['r band']
```

```python
# galaxy 1

# grab the columns we want
data1 = pd.read_csv('../Data/individual_analogs/453134_star_properties.csv') 
print(data1['g band'])
print(data1['r band'])

#create a histogram of the color. label the blue and red side. title your plot the galaxy name. calculate the mean color
# Plot histogram of color
plt.figure(figsize=(10,6))
plt.hist(color, bins=30, color='gray', edgecolor='black')

# Label the plot
plt.xlabel('Color (g - r)')
plt.ylabel('Number of Stars')
plt.title('Galaxy 453134 Color Distribution')

# Add vertical line for mean color
mean_color = np.mean(color)
plt.axvline(mean_color, color='red', linestyle='dashed', linewidth=2, label=f'Mean Color = {mean_color:.2f}')

# Annotate red/blue side
plt.text(mean_color + 0.1, plt.ylim()[1]*0.8, 'Redder', color='red')
plt.text(mean_color - 0.5, plt.ylim()[1]*0.8, 'Bluer', color='blue')

plt.legend()
plt.grid(True)
plt.show()

# Print the mean color
print(f"Mean color (g - r) for Galaxy 453134: {mean_color:.2f}")
```

```python
# galaxy 1

import matplotlib.pyplot as plt
import matplotlib
matplotlib.rc('font', **{'family':'sans-serif', 'size':16}) 
import numpy as np

# grab the columns we want
r1 = data1['r band']
g1 = data1['g band']

# calculate the color
color1 = g1 - r1

plt.hist(color1, bins=100)
plt.axvline(0.6, color='black')
plt.annotate('bluer', (0.1, 4000), color='blue', fontsize=16)
plt.annotate('redder', (.9, 4000), color='red', fontsize=16)
plt.xlim(-.1, 1.1)
plt.xlabel('Color [g-r]')
plt.ylabel('Number of Galaxies')
plt.title('460823')
print('The mean color is', round(np.mean(color1), 2))
```
