## Density of Stars

Let's now explore the densities of stars. If you recall from STP, density is the amount of stuff something has (the mass) in a given space (the volume). **Looking at the data table, what variables might be helpful for us to calculate the density of stars?** 

**Answer:**

The equation for density is $\ \rm D = \frac{Mass}{Volume}$. Stars are spheres whose volumes can be calculated using the equation $\ \rm V = \frac{4}{3}\pi r^3$, where r is the radius and $\pi$ is 3.14

```python
# create variables for the mass and radius

# Define the mass and radius

mass = data['Mass [Solar Mass]']
radius = data['Radius [Solar Radius]']
age = data['Age [Gyr]']
```

```python
# calculate the volume of the stars below. try doing so with a function!
# the resulting units will be (Ro)^3

# Calculate the volume using the formula v = 4/3 * 3.14 * r^3
volume = (4/3) * 3.14 * radius**3
density = mass/volume
print(volume)
print(density)
```

## Sun Density

Since a directly proportional model won't describe the data well, let's calculate the density of the Sun. Remember density is $\ \rm D = \frac{Mass}{Volume}$  . Make sure your density is g/mL

```python
# calculate the density of the Sun below in g/mL

# Function to convert mass to grams (g)
def convert_mass_to_grams(mass):
    return mass * (1.989e30) * 1000

# Function to convert volume to milliliters (mL)
def convert_volume_to_milliliters(volume):
    return volume * (695700 ** 3) * (100000 ** 3)

# Define the Sun's data
sun_mass_kg = 1.989e30  # Mass of the Sun in kilograms
sun_radius_m = 6.96e8  # Radius of the Sun in meters

# Calculate the Sun's volume in cubic meters
sun_volume_m3 = (4/3) * 3.14 * sun_radius_m**3

# Convert the Sun's mass to grams
sun_mass_g = convert_mass_to_grams(sun_mass_kg)

# Convert the Sun's volume to milliliters
sun_volume_ml = convert_volume_to_milliliters(sun_volume_m3)

# Calculate the density of the Sun in g/mL
sun_density_g_per_ml = sun_mass_g / sun_volume_ml

print(f"Density of the Sun: {sun_density_g_per_ml} g/mL")
```
